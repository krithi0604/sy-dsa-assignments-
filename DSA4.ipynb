```python

# Assignment 5: Bubble Sort and Insertion Sort
# Sorting Employee Records by Salary

# Bubble Sort
def bubble_sort(employees):
    arr = employees.copy()
    n = len(arr)
    comparisons = 0
    swaps = 0

    for i in range(n - 1):
        swapped = False

        for j in range(n - i - 1):
            comparisons += 1

            if arr[j]["salary"] > arr[j + 1]["salary"]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swaps += 1
                swapped = True

        if not swapped:
            break

    return arr, comparisons, swaps


# Insertion Sort
def insertion_sort(employees):
    arr = employees.copy()
    comparisons = 0
    shifts = 0

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if arr[j]["salary"] > key["salary"]:
                arr[j + 1] = arr[j]
                shifts += 1
                j -= 1
            else:
                break

        arr[j + 1] = key

    return arr, comparisons, shifts


# Display employee records
def display(employees):
    print("\nID\tName\t\tSalary")
    print("-" * 35)

    for emp in employees:
        print(f'{emp["id"]}\t{emp["name"]:<12}\t{emp["salary"]}')


# Main Program
employees = []

n = int(input("Enter number of employees: "))

for i in range(n):
    print("\nEmployee", i + 1)

    emp_id = input("Enter employee ID: ")
    name = input("Enter employee name: ")
    salary = float(input("Enter employee salary: "))

    employees.append({
        "id": emp_id,
        "name": name,
        "salary": salary
    })

print("\nOriginal Employee Records:")
display(employees)

# Bubble Sort
bubble, b_comp, b_swaps = bubble_sort(employees)

print("\n--- Bubble Sort: Salary in Ascending Order ---")
display(bubble)
print("Comparisons:", b_comp)
print("Swaps:", b_swaps)
print("Time Complexity: Best O(n), Average/Worst O(n^2)")
print("Space Complexity: O(n) including copied list")

# Insertion Sort
insertion, i_comp, i_shifts = insertion_sort(employees)

print("\n--- Insertion Sort: Salary in Ascending Order ---")
display(insertion)
print("Comparisons:", i_comp)
print("Shifts:", i_shifts)
print("Time Complexity: Best O(n), Average/Worst O(n^2)")
print("Space Complexity: O(n) including copied list")
```
