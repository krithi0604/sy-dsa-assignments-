```python

# Assignment 4: Linear Search and Binary Search

# Linear Search
def linear_search(arr, key):
    comparisons = 0

    for i in range(len(arr)):
        comparisons += 1

        if arr[i] == key:
            return i, comparisons

    return -1, comparisons


# Binary Search
def binary_search(arr, key):
    low = 0
    high = len(arr) - 1
    comparisons = 0

    while low <= high:
        mid = (low + high) // 2
        comparisons += 1

        if arr[mid] == key:
            return mid, comparisons
        elif arr[mid] < key:
            low = mid + 1
        else:
            high = mid - 1

    return -1, comparisons


# Main Program
arr = list(map(int, input(
    "Enter numbers separated by spaces: "
).split()))

key = int(input("Enter number to search: "))

# Linear Search
index, count = linear_search(arr, key)

print("\n--- Linear Search ---")
if index != -1:
    print("Element found at position:", index + 1)
else:
    print("Element not found")

print("Comparisons:", count)
print("Time Complexity: O(n)")
print("Space Complexity: O(1)")

# Binary Search requires a sorted list
sorted_arr = sorted(arr)

print("\nSorted list for Binary Search:", sorted_arr)

index, count = binary_search(sorted_arr, key)

print("\n--- Binary Search ---")
if index != -1:
    print("Element found at position:", index + 1)
else:
    print("Element not found")

print("Comparisons:", count)
print("Best/Average Time: O(log n)")
print("Worst Time: O(log n)")
print("Space Complexity: O(1)")
```
