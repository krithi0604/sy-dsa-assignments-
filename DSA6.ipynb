```python

# Assignment 9: Product Catalog using AVL Tree

class Node:
    def __init__(self, product_id, name, price):
        self.product_id = product_id
        self.name = name
        self.price = price
        self.left = None
        self.right = None
        self.height = 1


class AVLTree:
    def height(self, node):
        if node is None:
            return 0
        return node.height

    def balance_factor(self, node):
        if node is None:
            return 0
        return self.height(node.left) - self.height(node.right)

    def update_height(self, node):
        node.height = 1 + max(
            self.height(node.left),
            self.height(node.right)
        )

    # Right rotation
    def right_rotate(self, y):
        x = y.left
        temp = x.right

        x.right = y
        y.left = temp

        self.update_height(y)
        self.update_height(x)

        return x

    # Left rotation
    def left_rotate(self, x):
        y = x.right
        temp = y.left

        y.left = x
        x.right = temp

        self.update_height(x)
        self.update_height(y)

        return y

    # Insert a product
    def insert(self, root, product_id, name, price):
        if root is None:
            return Node(product_id, name, price)

        if product_id < root.product_id:
            root.left = self.insert(
                root.left, product_id, name, price
            )
        elif product_id > root.product_id:
            root.right = self.insert(
                root.right, product_id, name, price
            )
        else:
            print("Product ID already exists!")
            return root

        self.update_height(root)
        balance = self.balance_factor(root)

        # Left Left case
        if balance > 1 and product_id < root.left.product_id:
            return self.right_rotate(root)

        # Right Right case
        if balance < -1 and product_id > root.right.product_id:
            return self.left_rotate(root)

        # Left Right case
        if balance > 1 and product_id > root.left.product_id:
            root.left = self.left_rotate(root.left)
            return self.right_rotate(root)

        # Right Left case
        if balance < -1 and product_id < root.right.product_id:
            root.right = self.right_rotate(root.right)
            return self.left_rotate(root)

        return root

    # Search a product and count comparisons
    def search(self, root, product_id):
        comparisons = 0
        current = root

        while current is not None:
            comparisons += 1

            if product_id == current.product_id:
                return current, comparisons
            elif product_id < current.product_id:
                current = current.left
            else:
                current = current.right

        return None, comparisons

    # Find the smallest node
    def min_value_node(self, node):
        current = node

        while current.left is not None:
            current = current.left

        return current

    # Delete a product
    def delete(self, root, product_id):
        if root is None:
            return root

        if product_id < root.product_id:
            root.left = self.delete(root.left, product_id)
        elif product_id > root.product_id:
            root.right = self.delete(root.right, product_id)
        else:
            # Node with one child or no child
            if root.left is None:
                return root.right
            elif root.right is None:
                return root.left

            # Node with two children
            successor = self.min_value_node(root.right)

            root.product_id = successor.product_id
            root.name = successor.name
            root.price = successor.price

            root.right = self.delete(
                root.right, successor.product_id
            )

        if root is None:
            return root

        self.update_height(root)
        balance = self.balance_factor(root)

        # Left Left case
        if balance > 1 and self.balance_factor(root.left) >= 0:
            return self.right_rotate(root)

        # Left Right case
        if balance > 1 and self.balance_factor(root.left) < 0:
            root.left = self.left_rotate(root.left)
            return self.right_rotate(root)

        # Right Right case
        if balance < -1 and self.balance_factor(root.right) <= 0:
            return self.left_rotate(root)

        # Right Left case
        if balance < -1 and self.balance_factor(root.right) > 0:
            root.right = self.right_rotate(root.right)
            return self.left_rotate(root)

        return root

    # Update product details
    def update_product(self, root, product_id):
        product, comparisons = self.search(root, product_id)

        print("Comparisons:", comparisons)

        if product is None:
            print("Product not found!")
            return

        name = input("Enter new product name: ")
        price = float(input("Enter new price: "))

        product.name = name
        product.price = price

        print("Product updated successfully!")

    # Display products in sorted order
    def display(self, root):
        if root is not None:
            self.display(root.left)
            print(
                "ID:", root.product_id,
                "| Name:", root.name,
                "| Price:", root.price
            )
            self.display(root.right)


# Main program
tree = AVLTree()
root = None

while True:
    print("\n===== Product Catalog =====")
    print("1. Add Product")
    print("2. Search Product")
    print("3. Update Product")
    print("4. Delete Product")
    print("5. Display Products")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        product_id = int(input("Enter product ID: "))
        name = input("Enter product name: ")
        price = float(input("Enter product price: "))

        old_root = root
        root = tree.insert(root, product_id, name, price)

        if root is not old_root or tree.search(root, product_id)[0]:
            # The tree insertion function handles duplicate IDs.
            pass

    elif choice == "2":
        product_id = int(input("Enter product ID to search: "))
        product, comparisons = tree.search(root, product_id)

        print("Comparisons:", comparisons)

        if product is not None:
            print("Product found!")
            print("ID:", product.product_id)
            print("Name:", product.name)
            print("Price:", product.price)
        else:
            print("Product not found!")

    elif choice == "3":
        product_id = int(input("Enter product ID to update: "))
        tree.update_product(root, product_id)

    elif choice == "4":
        product_id = int(input("Enter product ID to delete: "))
        product, comparisons = tree.search(root, product_id)

        if product is None:
            print("Product not found!")
        else:
            root = tree.delete(root, product_id)
            print("Product deleted successfully!")

    elif choice == "5":
        if root is None:
            print("Product catalog is empty!")
        else:
            print("\nProducts sorted by ID:")
            tree.display(root)

    elif choice == "6":
        print("Exiting Product Catalog.")
        break

    else:
        print("Invalid choice!")
```
