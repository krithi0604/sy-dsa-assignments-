```python

# Appointment Management System using List

appointments = []

# Add a new appointment
def add_appointment():
    name = input("Enter patient name: ")
    date = input("Enter appointment date: ")
    time = input("Enter appointment time: ")

    appointment = [name, date, time]
    appointments.append(appointment)

    print("Appointment added successfully!")


# Insert appointment at a specific position
def insert_appointment():
    name = input("Enter patient name: ")
    date = input("Enter appointment date: ")
    time = input("Enter appointment time: ")

    appointment = [name, date, time]

    pos = int(input("Enter position to insert: "))

    if 1 <= pos <= len(appointments) + 1:
        appointments.insert(pos - 1, appointment)
        print("Appointment inserted successfully!")
    else:
        print("Invalid position!")


# Delete an appointment by patient name
def delete_appointment():
    name = input("Enter patient name to delete: ")

    for appointment in appointments:
        if appointment[0].lower() == name.lower():
            appointments.remove(appointment)
            print("Appointment deleted successfully!")
            return

    print("Appointment not found!")


# Search for an appointment by patient name
def search_appointment():
    name = input("Enter patient name to search: ")

    for appointment in appointments:
        if appointment[0].lower() == name.lower():
            print("Appointment found:")
            print("Patient:", appointment[0])
            print("Date:", appointment[1])
            print("Time:", appointment[2])
            return

    print("Appointment not found!")


# Display all appointments
def display_appointments():
    if not appointments:
        print("No appointments available.")
        return

    print("\n--- Appointment List ---")

    for i, appointment in enumerate(appointments, start=1):
        print(i, appointment[0],
              "| Date:", appointment[1],
              "| Time:", appointment[2])


# Main menu
while True:
    print("\n===== Appointment Management =====")
    print("1. Add Appointment")
    print("2. Insert Appointment")
    print("3. Delete Appointment")
    print("4. Search Appointment")
    print("5. Display Appointments")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_appointment()
    elif choice == "2":
        insert_appointment()
    elif choice == "3":
        delete_appointment()
    elif choice == "4":
        search_appointment()
    elif choice == "5":
        display_appointments()
    elif choice == "6":
        print("Exiting Appointment Management System.")
        break
    else:
        print("Invalid choice! Try again.")
```
