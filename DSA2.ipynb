```python
###### # Book Inventory using Singly Linked List

# Node class
class Node:
    def __init__(self, book_id, title):
        self.book_id = book_id
        self.title = title
        self.next = None


# Linked List class
class BookInventory:
    def __init__(self):
        self.head = None

    # Add a book at the end
    def add_book(self):
        book_id = input("Enter book ID: ")
        title = input("Enter book title: ")

        new_node = Node(book_id, title)

        if self.head is None:
            self.head = new_node
        else:
            temp = self.head
            while temp.next is not None:
                temp = temp.next
            temp.next = new_node

        print("Book added successfully!")

    # Delete a book using its ID
    def delete_book(self):
        book_id = input("Enter book ID to delete: ")

        if self.head is None:
            print("Inventory is empty!")
            return

        # If the first node is to be deleted
        if self.head.book_id == book_id:
            self.head = self.head.next
            print("Book deleted successfully!")
            return

        temp = self.head

        while temp.next is not None:
            if temp.next.book_id == book_id:
                temp.next = temp.next.next
                print("Book deleted successfully!")
                return
            temp = temp.next

        print("Book not found!")

    # Count total books
    def count_books(self):
        count = 0
        temp = self.head

        while temp is not None:
            count += 1
            temp = temp.next

        print("Total books:", count)

    # Display all books
    def display_books(self):
        if self.head is None:
            print("Inventory is empty!")
            return

        temp = self.head
        print("\n--- Book Inventory ---")

        while temp is not None:
            print("Book ID:", temp.book_id,
                  "| Title:", temp.title)
            temp = temp.next


# Main program
inventory = BookInventory()

while True:
    print("\n===== Book Inventory Menu =====")
    print("1. Add Book")
    print("2. Delete Book")
    print("3. Count Books")
    print("4. Display Books")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        inventory.add_book()
    elif choice == "2":
        inventory.delete_book()
    elif choice == "3":
        inventory.count_books()
    elif choice == "4":
        inventory.display_books()
    elif choice == "5":
        print("Exiting Book Inventory System.")
        break
    else:
        print("Invalid choice! Try again.")
```
